```python
import rclpy
from std_msgs.msg import String
```

```python
rclpy.init()
node = rclpy.create_node('ros_node')
pub = node.create_publisher(String, 'chatter', 10)
```

```python
msg = String()
msg.data = "Hello, ROS"
pub.publish(msg)
rclpy.spin_once(node, timeout_sec = 0.1)
node.destroy_node()
rclpy.shutdown()
```
