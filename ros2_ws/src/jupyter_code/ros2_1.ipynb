```python
import rclpy
from std_msgs.msg import String
```

```python
rclpy.init()
node = rclpy.create_node('nb_node')
pub = node.create_publisher(String, 'nb', 10)
```

```python
msg = String()
msg.data = "Hello, ROS"
pub.publish(msg)
rclpy.spin_once(node, timeout_sec = 0.1)
# rclpy.spin(node)
```

```python
node.destroy_node()
rclpy.shutdown()
```
