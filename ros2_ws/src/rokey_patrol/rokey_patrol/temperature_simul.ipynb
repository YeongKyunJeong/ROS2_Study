```python
import random
import matplotlib.pyplot as plt
```

```python
thresh = 0.1
sensor_data = []
for i in range(100):
    if random.random() < thresh:
        sensor_data.append(random.uniform(50, 60))
    else:
        # sensor_data.append(25+random.uniform(-1, 1))
        sensor_data.append(random.uniform(24, 26))
```

```python
plt.plot(sensor_data)
```
